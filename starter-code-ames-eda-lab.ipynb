## Part 7 — Covariance and Correlation

Use covariance and correlation to investigate relationships between numeric variables.


```python
housing.select_dtypes(include='number').corr()
```

The numeric values that have teh highest correlation is Gr_Liv_Area with a 0.706780 correlation 

```python
housing[['Gr_Liv_Area','Sale_Price']].cov()
```

```python
y=housing["Sale_Price"]
x=housing["Gr_Liv_Area"]
```

```python
mean_x = x.mean()
mean_y = y.mean()
```

```python
cov = sum((xi - mean_x)*(yi - mean_y) for xi, yi in zip(x,y)) / len(x)

print(f" The covariance of Gr_Liv_Area and Sale_price is {cov}")
```

```python
corr = cov / (x.std() * y.std())
print(f"The correlation between Gr_Liv_Area and Sale_price is: {corr}")
```
